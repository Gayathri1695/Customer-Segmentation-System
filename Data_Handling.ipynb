```python
import pandas as pd
df["Age"]=df["Age"].fillna(df["Age"].mean())
df["Annual_Income"]=df["Annual_Income"].fillna(df["Annual_Income"].mean())
df["Spending_Score"]=df["Spending_Score"].fillna(df["Spending_Score"].mean())
display(df)
```

```python
df = df.dropna()
```
