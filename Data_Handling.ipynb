```python

X = df[[
    "Age",
    "Annual_Income",
    "Website_Visits",
    "Purchase_Frequency"
]]

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)


df["Cluster"] = model.fit_predict(X_scaled)

print("\nCLUSTER ANALYSIS\n")

cluster_summary = df.groupby("Cluster")[[
    "Age",
    "Annual_Income",
    "Website_Visits",
    "Purchase_Frequency"
]].mean().astype(int)

print(cluster_summary)


premium_cluster = cluster_summary["Purchase_Frequency"].idxmax()

inactive_cluster = cluster_summary["Purchase_Frequency"].idxmin()

regular_cluster = [cluster for cluster in cluster_summary.index
    if cluster not in [premium_cluster, inactive_cluster]][0]


segment_names = {
    premium_cluster: "Premium Customer",
    regular_cluster: "Regular Customer",
    inactive_cluster: "Inactive Customer"
}


df["Customer_Segment"] = df["Cluster"].map(segment_names)



print("\n CUSTOMER SEGMENTS \n")

result = df[[
    "Customer_ID",
    "Age",
    "Annual_Income",
    "Website_Visits",
    "Purchase_Frequency",
    "Cluster",
    "Customer_Segment"
]]
print(result.head(20).to_string(index=False))


print("\n SEGMENT SUMMARY \n")

for cluster, segment in segment_names.items():

    print("Cluster", cluster, "=", segment)

    customers = df[df["Cluster"] == cluster]

    print("Number of Customers:", len(customers))

    print()
```

```python

new_customer = [[30, 60000, 5, 8]]
new_customer_scaled = scaler.transform(new_customer)
prediction = model.predict(new_customer_scaled)
cluster_number = prediction[0]
customer_segment = segment_names[cluster_number]

print("\nNEW CUSTOMER PREDICTION:\n")

print("Age:", new_customer[0][0])
print("Annual Income:", new_customer[0][1])
print("Website Visits:", new_customer[0][2])
print("Purchase Frequency:", new_customer[0][3])

print("\nPredicted Cluster:", cluster_number)
print("Predicted Customer Segment:", customer_segment)
```
